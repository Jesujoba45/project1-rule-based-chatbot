```python
responses = {
    "hello": "Hi there! Nice to meet you.",
    "hi": "Hello friend!",
    "how are you": "I am doing great. Thanks for asking!",
    "what is your name": "I am DecodeBot, your AI assistant.",
    "thanks": "You are welcome!",
    "who created you": "I was created as an internship AI project.",
    "bye": "Goodbye! Have a great day!"
}

print("=== Welcome to DecodeBot ===")
print("Type 'exit' anytime to end the chat.\n")

while True:
    user_input = input("You: ").lower().strip()

    if user_input == "exit":
        print("Bot: Goodbye! Thanks for chatting.")
        break

    response = responses.get(
        user_input,
        "Bot: Sorry, I don't understand that."
    )

    print(response)
```
